```python
from ToneDS import ToneDS
from torch.utils.data import DataLoader

ds = ToneDS("BZNSYP/Wave", "BZNSYP/PhoneLabeling")

loader = DataLoader(ds, batch_size=1, shuffle=True)

for mel, tone in loader:
    print("Mel shape:", mel.shape)   # [1, 1, 80, 100]
    print("Tone:", tone.item())
    break
```

```python
import matplotlib.pyplot as plt

mel, tone = ds[0]
mel_np = mel.squeeze(0).numpy()
tone_int = int(tone)

plt.figure(figsize=(8, 4))
plt.imshow(mel_np)
plt.title(f"Tone: {tone_int + 1}")
plt.xlabel("Time")
plt.ylabel("Mel-freq")
plt.colorbar(label="dB")
plt.show()
```

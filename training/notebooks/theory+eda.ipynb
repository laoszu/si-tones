## Chinese Phonology
I will be focusing on the Standard Mandarin Chinese 普通话 (pǔ tōng huà) pronunciation, which originates from the Beijing dialect. Mandarin in a tonal languages and its four main tones are:
* First Tone (阴平 yīnpíng), is the high level tone with a pitch contour of 55. Represented by the symbol ˉ as in mā (妈), it maintains a high, flat, and steady pitch throughout the syllable duration;
* Second Tone (阳平 yángpíng), is the rising tone with a pitch contour of 35. Marked by the symbol ˊ as in má (麻), it features a medium to high rising pitch pattern.
* Third Tone (上声 shǎngshēng), is the dipping tone with a complex pitch contour of 214. Indicated by the symbol ˇ as in mǎ (马), it demonstrates a characteristic low falling-rising contour that dips before rising.
* Fourth Tone (去声 qùshēng), is the falling tone with a pitch contour of 51. Denoted by the symbol ˋ as in mà (骂), it exhibits a sharp high to low falling pattern.
* Neutral Tone (轻声 qīngshēng), is sometimes thought of as a lack of tone. It is associated with weak syllables, which are generally somewhat shorter than tonic syllables. <br/>

It's worth to mention, that it's not the same as the taiwaneese mandarin 國語 (guó yǔ) which has a slightly different pronunciation niuances, such as different retroflex consonants like
zh-, ch-, and sh-, the third tone often has no rise and a lack of erization (儿化).

## Acoustic Analysis of Speech
The purpose of the acoustic analysis is to change the audio signal into a numerical data. Data in the `data_thchs30` is the wav format, which represents the intensivity of a sound in the domian of time.

## Fundamental Frequency ($f_0$)
The objects of my interest is the fundamental frequency ($f_0$), represents the pitch contour of speech and is the primary acoustic correlate of Mandarin tones. It corresponds to the vibration frequency of the vocal folds and is extracted using probabilistic YIN (PYIN) algorithm:<br/>
$f_0=\frac{1}{τ}$ where τ maximizes $r(τ)=\sum x(t)⋅x(t+τ)$

For Mandarin Chinese, the typical $f_0$ ranges are:
- Tone 1: 155-165 Hz (high level)
- Tone 2: 120-220 Hz (rising)
- Tone 3: 90-180-140 Hz (dipping)
- Tone 4: 220-90 Hz (falling)

## Formant Frequencies ($f_1$, $f_2$, $f_3$)

Formants are resonance frequencies of the vocal tract that characterize vowel sounds. They are extracted using Linear Predictive Coding (LPC):

$$x[n]=\sum_{k=1}^{p} a_{k}x[n−k]+e[n] $$

where $a_k$ are LPC coefficients and $e[n]$ is the excitation signal. The formant frequencies are obtained by solving:

$$1−\sum_{k=1}^{p}a_{k}z^{-1}=0$$

Typical formant values for Mandarin vowels:
- Vowel /a/: $f_1$ ≈ 800 Hz, $f_2$ ≈ 1200 Hz
- Vowel /i/: $f_1$ ≈ 300 Hz, $f_2$ ≈ 2300 Hz
- Vowel /u/: $f_1$ ≈ 300 Hz, $f_2$ ≈ 800 Hz
